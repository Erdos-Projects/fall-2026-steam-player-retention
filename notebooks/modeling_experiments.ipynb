```python
import pandas as pd
import numpy as np
from pathlib import Path
import re
raw_dir = Path(r"C:\Projects\data-science\data\raw")

# Load Steam application metadata
applications = pd.read_csv(
    raw_dir / "steam_dataset_2025_csv" / "applications.csv",
    low_memory=False
)

# Keep actual games
games = applications[
    applications["type"].eq("game")
].copy()

# Parse dates
games["release_date"] = pd.to_datetime(
    games["release_date"],
    errors="coerce"
)

games["release_year"] = games["release_date"].dt.year

# Keep the period we are studying
games = games[
    games["release_year"].between(2015, 2024)
].copy()

print("Steam games:", len(games))
steamcharts_path = Path(
    r"C:\Projects\data-science\data\raw\steamcharts.csv"
)

steamcharts = pd.read_csv(
    steamcharts_path,
    low_memory=False
)

print(steamcharts.shape)
print(steamcharts.columns.tolist())
steamcharts.head()

print("Unique games:", steamcharts["steam_appid"].nunique())

steamcharts["date"] = pd.to_datetime(
    steamcharts["month"],
    format="%b-%y"
)

print("Earliest month:", steamcharts["date"].min())
print("Latest month:", steamcharts["date"].max())


print(
    "Unique Steam games:",
    steamcharts["steam_appid"].nunique()
)

steamcharts = steamcharts.rename(
    columns={"steam_appid": "appid"}
)

common_appids = (
    set(games["appid"])
    & set(steamcharts["appid"])
)

print(
    "Games in SteamCharts:",
    steamcharts["appid"].nunique()
)

print(
    "Games overlapping Steam metadata:",
    len(common_appids)
)

print(
    "Duplicate game-month rows:",
    steamcharts.duplicated(
        subset=["appid", "date"]
    ).sum()
)

# Make sure dates are parsed
steamcharts["date"] = pd.to_datetime(
    steamcharts["month"],
    format="%b-%y"
)

steamcharts["month_period"] = (
    steamcharts["date"].dt.to_period("M")
)

# Keep only games appearing in both datasets
games_with_history = games[
    games["appid"].isin(common_appids)
].copy()

# Define release month
games_with_history["release_month"] = (
    games_with_history["release_date"]
    .dt.to_period("M")
)

# First full month after release
games_with_history["month_1"] = (
    games_with_history["release_month"] + 1
)

# Month approximately one year after release
games_with_history["month_12"] = (
    games_with_history["release_month"] + 12
)


month1 = steamcharts[
    ["appid", "month_period", "avg_players", "peak_players"]
].rename(
    columns={
        "month_period": "month_1",
        "avg_players": "month1_avg_players",
        "peak_players": "month1_peak_players"
    }
)

month12 = steamcharts[
    ["appid", "month_period", "avg_players", "peak_players"]
].rename(
    columns={
        "month_period": "month_12",
        "avg_players": "month12_avg_players",
        "peak_players": "month12_peak_players"
    }
)


model_large = games_with_history.merge(
    month1,
    on=["appid", "month_1"],
    how="left"
)

model_large = model_large.merge(
    month12,
    on=["appid", "month_12"],
    how="left"
)

print("Overlapping games:", len(model_large))

print(
    "Month 1 available:",
    model_large["month1_avg_players"].notna().sum()
)

print(
    "Month 12 available:",
    model_large["month12_avg_players"].notna().sum()
)

print(
    "Both available:",
    model_large[
        ["month1_avg_players", "month12_avg_players"]
    ].notna().all(axis=1).sum()
)

usable = model_large.dropna(
    subset=[
        "month1_avg_players",
        "month12_avg_players"
    ]
).copy()

print(
    usable["release_year"]
    .value_counts()
    .sort_index()
)

usable["month1_avg_players"].describe(
    percentiles=[0.01, 0.05, 0.10, 0.25, 0.50]
)

for threshold in [0.5, 1, 2, 5, 10]:
    n = (usable["month1_avg_players"] >= threshold).sum()

    print(
        f"Threshold >= {threshold}: "
        f"{n} games "
        f"({n / len(usable):.1%})"
    )



    # Main modeling target: uses all games
usable["player_change"] = (
    np.arcsinh(usable["month12_avg_players"])
    - np.arcsinh(usable["month1_avg_players"])
)

usable["retention_ratio"] = np.where(
    usable["month1_avg_players"] > 0,
    usable["month12_avg_players"]
    / usable["month1_avg_players"],
    np.nan
)

usable["player_change"].describe()

usable["asinh_player_change"] = (
    np.arcsinh(usable["month12_avg_players"])
    - np.arcsinh(usable["month1_avg_players"])
)


target = "asinh_player_change"


usable.groupby("release_year")[
    "asinh_player_change"
].agg(["count", "mean", "median", "std"])


development_df = usable[
    usable["release_year"] <= 2021
].copy()

test_df = usable[
    usable["release_year"].between(2022, 2023)
].copy()

partial_2024_df = usable[
    usable["release_year"] == 2024
].copy()

print("Development:", len(development_df))
print("Final test:", len(test_df))
print("Partial 2024:", len(partial_2024_df))

usable["years_since_2015"] = (
    usable["release_year"] - 2015
)

# ============================================================
# PATHS
# ============================================================

PROJECT_DIR = Path(r"C:\Projects\data-science")

STEAM_DATA_DIR = (
    PROJECT_DIR
    / "data"
    / "raw"
    / "steam_dataset_2025_csv"
)

PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# 1. START FROM THE 4,200 USABLE GAMES
# ============================================================

model_df = usable.copy()

print("Starting games:", len(model_df))


# ============================================================
# 2. LANGUAGE COUNT
# ============================================================

def count_languages(x):
    if pd.isna(x):
        return 0

    # Remove explanatory text after <br>
    x = x.split("<br>")[0]

    # Remove HTML tags
    x = re.sub(r"<[^>]+>", "", x)

    languages = [
        lang.strip()
        for lang in x.split(",")
        if lang.strip()
    ]

    return len(languages)


model_df["language_count"] = (
    model_df["supported_languages"]
    .apply(count_languages)
)


# ============================================================
# 3. PRICE IN USD
# ============================================================

model_df["initial_price_usd"] = np.where(
    model_df["mat_currency"] == "USD",
    model_df["mat_initial_price"] / 100,
    np.nan
)


# ============================================================
# 4. TIME FEATURE
# ============================================================

model_df["years_since_2015"] = (
    model_df["release_year"] - 2015
)


# ============================================================
# 5. LOAD GENRES
# ============================================================

application_genres = pd.read_csv(
    STEAM_DATA_DIR / "application_genres.csv"
)

genres = pd.read_csv(
    STEAM_DATA_DIR / "genres.csv"
)

genre_map = application_genres.merge(
    genres,
    left_on="genre_id",
    right_on="id",
    how="left"
)

genre_map = genre_map[
    genre_map["appid"].isin(model_df["appid"])
].copy()


# ============================================================
# 6. REMOVE SOFTWARE-LIKE APPLICATIONS
# ============================================================

non_game_genres = [
    "Design & Illustration",
    "Utilities",
    "Game Development",
    "Education",
    "Video Production",
    "Software Training",
    "Audio Production",
    "Web Publishing",
    "Animation & Modeling",
]

non_game_appids = genre_map.loc[
    genre_map["name"].isin(non_game_genres),
    "appid"
].unique()

model_df = model_df[
    ~model_df["appid"].isin(non_game_appids)
].copy()

print(
    "After removing software-like applications:",
    len(model_df)
)


# Keep genre records only for surviving games
genre_map = genre_map[
    genre_map["appid"].isin(model_df["appid"])
].copy()


# ============================================================
# 7. SELECT MAIN GAME GENRES
# ============================================================

main_genres = [
    "Action",
    "Adventure",
    "Casual",
    "Early Access",
    "Free To Play",
    "Indie",
    "Massively Multiplayer",
    "RPG",
    "Racing",
    "Simulation",
    "Sports",
    "Strategy",
]

genre_subset = genre_map[
    genre_map["name"].isin(main_genres)
].copy()


# ============================================================
# 8. ONE-HOT ENCODE GENRES
# ============================================================

genre_features = pd.crosstab(
    genre_subset["appid"],
    genre_subset["name"]
)

genre_features = (
    genre_features > 0
).astype(int)

genre_features.columns = [
    "genre_" + col.replace(" ", "_")
    for col in genre_features.columns
]

model_df = model_df.merge(
    genre_features,
    left_on="appid",
    right_index=True,
    how="left"
)

genre_columns = [
    col
    for col in model_df.columns
    if col.startswith("genre_")
]

model_df[genre_columns] = (
    model_df[genre_columns]
    .fillna(0)
    .astype(int)
)


# ============================================================
# 9. CREATE MODELING TARGET
# ============================================================

model_df["asinh_player_change"] = (
    np.arcsinh(model_df["month12_avg_players"])
    - np.arcsinh(model_df["month1_avg_players"])
)


# Keep raw retention ratio for interpretation only
model_df["retention_ratio"] = np.where(
    model_df["month1_avg_players"] > 0,
    model_df["month12_avg_players"]
    / model_df["month1_avg_players"],
    np.nan
)


# ============================================================
# 10. OPTIONAL TRANSFORM OF FIRST-MONTH POPULARITY
# ============================================================

# Much easier for models to handle than raw player counts
model_df["asinh_month1_players"] = np.arcsinh(
    model_df["month1_avg_players"]
)


# ============================================================
# 11. DEFINE MODEL FEATURES
# ============================================================

feature_cols = [
    "is_free",
    "initial_price_usd",
    "language_count",
    "years_since_2015",
    "mat_supports_windows",
    "mat_supports_mac",
    "mat_supports_linux",
    "asinh_month1_players",
] + genre_columns

target = "asinh_player_change"


# ============================================================
# 12. CREATE CLEAN MODELING TABLE
# ============================================================

id_cols = [
    "appid",
    "name",
    "release_date",
    "release_year",
]

diagnostic_cols = [
    "month1_avg_players",
    "month12_avg_players",
    "retention_ratio",
]

modeling_columns = (
    id_cols
    + feature_cols
    + diagnostic_cols
    + [target]
)

modeling_table = model_df[
    modeling_columns
].copy()


# ============================================================
# 13. INSPECT TABLE
# ============================================================

print("\nModeling table shape:")
print(modeling_table.shape)

print("\nFeatures:")
for col in feature_cols:
    print(col)

print("\nTarget:")
print(target)

print("\nMissing values:")
print(
    modeling_table[feature_cols]
    .isna()
    .sum()
    .sort_values(ascending=False)
)

print("\nRelease-year distribution:")
print(
    modeling_table["release_year"]
    .value_counts()
    .sort_index()
)

display(modeling_table.head())


# ============================================================
# 14. SAVE TABLE
# ============================================================

OUTPUT_PATH = (
    PROCESSED_DIR
    / "steam_modeling_data_large.csv"
)

modeling_table.to_csv(
    OUTPUT_PATH,
    index=False
)

model_df[
    ["month1_avg_players", "asinh_month1_players"]
].head(20)

print("\nSaved to:")
print(OUTPUT_PATH)


import pandas as pd
import numpy as np
from pathlib import Path

from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.ensemble import GradientBoostingRegressor
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)


# ============================================================
# 1. PATHS
# ============================================================

PROJECT_DIR = Path(r"C:\Projects\data-science")

DATA_PATH = (
    PROJECT_DIR
    / "data"
    / "processed"
    / "steam_modeling_data_large.csv"
)

RESULTS_DIR = PROJECT_DIR / "results"
RESULTS_DIR.mkdir(parents=True, exist_ok=True)


# ============================================================
# 2. LOAD MODELING DATA
# ============================================================

modeling_table = pd.read_csv(DATA_PATH)

print("Full modeling table shape:")
print(modeling_table.shape)


# ============================================================
# 3. DEFINE FEATURES AND TARGET
# ============================================================

genre_columns = [
    col
    for col in modeling_table.columns
    if col.startswith("genre_")
]

feature_cols = [
    "is_free",
    "initial_price_usd",
    "language_count",
    "years_since_2015",
    "mat_supports_windows",
    "mat_supports_mac",
    "mat_supports_linux",
    "asinh_month1_players",
] + genre_columns

target = "asinh_player_change"

print("\nNumber of features:")
print(len(feature_cols))

print("\nTarget:")
print(target)


# ============================================================
# 4. DEVELOPMENT / FINAL TEST SPLIT
# ============================================================

# Development data are used for model selection and tuning.
development_df = modeling_table[
    modeling_table["release_year"] <= 2021
].copy()

# Final holdout.
# DO NOT use this during hyperparameter tuning.
test_df = modeling_table[
    modeling_table["release_year"].between(2022, 2023)
].copy()

# 2024 is kept separately because it is an incomplete cohort.
partial_2024_df = modeling_table[
    modeling_table["release_year"] == 2024
].copy()

print("\nDevelopment set:")
print(len(development_df))

print("\nFinal 2022-2023 holdout:")
print(len(test_df))

print("\nPartial 2024 cohort:")
print(len(partial_2024_df))

print("\nMissing final test targets:")
print(test_df[target].isna().sum())


# ============================================================
# 5. TEMPORAL CROSS-VALIDATION FOLDS
# ============================================================

# A one-year gap is left between training and validation
# because the target requires approximately 12 months
# after release to become observable.

fold_definitions = [
    {
        "name": "Fold 1",
        "train_end": 2016,
        "val_year": 2018,
    },
    {
        "name": "Fold 2",
        "train_end": 2017,
        "val_year": 2019,
    },
    {
        "name": "Fold 3",
        "train_end": 2018,
        "val_year": 2020,
    },
    {
        "name": "Fold 4",
        "train_end": 2019,
        "val_year": 2021,
    },
]


print("\nTemporal fold sizes:")

for fold in fold_definitions:

    train_df = development_df[
        development_df["release_year"]
        <= fold["train_end"]
    ]

    val_df = development_df[
        development_df["release_year"]
        == fold["val_year"]
    ]

    print(
        fold["name"],
        "| Train:",
        len(train_df),
        "| Validation:",
        len(val_df)
    )


# ============================================================
# 6. REUSABLE TEMPORAL EVALUATION FUNCTION
# ============================================================

def evaluate_model_over_time(
    model_name,
    model,
    data,
    feature_cols,
    target,
    fold_definitions
):

    results = []

    for fold in fold_definitions:

        train_df = data[
            data["release_year"]
            <= fold["train_end"]
        ].copy()

        val_df = data[
            data["release_year"]
            == fold["val_year"]
        ].copy()

        X_train = train_df[feature_cols]
        y_train = train_df[target]

        X_val = val_df[feature_cols]
        y_val = val_df[target]

        # Imputation happens separately inside each
        # training fold to avoid leakage.
        pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "model",
                model
            )
        ])

        pipeline.fit(
            X_train,
            y_train
        )

        predictions = pipeline.predict(
            X_val
        )

        mae = mean_absolute_error(
            y_val,
            predictions
        )

        rmse = np.sqrt(
            mean_squared_error(
                y_val,
                predictions
            )
        )

        r2 = r2_score(
            y_val,
            predictions
        )

        results.append({
            "model": model_name,
            "fold": fold["name"],
            "train_end": fold["train_end"],
            "validation_year": fold["val_year"],
            "train_size": len(train_df),
            "validation_size": len(val_df),
            "MAE": mae,
            "RMSE": rmse,
            "R2": r2,
        })

    return pd.DataFrame(results)


# ============================================================
# 7. GRADIENT BOOSTING PARAMETER GRID
# ============================================================

# Keep this grid intentionally small.
# We want a controlled experiment rather than
# aggressively optimizing on only four validation cohorts.

gb_parameter_grid = [
    {
        "n_estimators": 100,
        "learning_rate": 0.05,
        "max_depth": 2,
    },
    {
        "n_estimators": 100,
        "learning_rate": 0.05,
        "max_depth": 3,
    },
    {
        "n_estimators": 100,
        "learning_rate": 0.10,
        "max_depth": 2,
    },
    {
        "n_estimators": 100,
        "learning_rate": 0.10,
        "max_depth": 3,
    },
    {
        "n_estimators": 200,
        "learning_rate": 0.05,
        "max_depth": 2,
    },
    {
        "n_estimators": 200,
        "learning_rate": 0.05,
        "max_depth": 3,
    },
]


# ============================================================
# 8. RUN GRADIENT BOOSTING EXPERIMENTS
# ============================================================

gb_tuning_results_list = []

for params in gb_parameter_grid:

    model_name = (
        f"GB_n{params['n_estimators']}"
        f"_lr{params['learning_rate']}"
        f"_d{params['max_depth']}"
    )

    print("\nRunning:", model_name)

    model = GradientBoostingRegressor(
        n_estimators=params["n_estimators"],
        learning_rate=params["learning_rate"],
        max_depth=params["max_depth"],
        random_state=42,
    )

    results = evaluate_model_over_time(
        model_name=model_name,
        model=model,
        data=development_df,
        feature_cols=feature_cols,
        target=target,
        fold_definitions=fold_definitions,
    )

    results["n_estimators"] = (
        params["n_estimators"]
    )

    results["learning_rate"] = (
        params["learning_rate"]
    )

    results["max_depth"] = (
        params["max_depth"]
    )

    gb_tuning_results_list.append(
        results
    )


# Combine all configurations
gb_tuning_results = pd.concat(
    gb_tuning_results_list,
    ignore_index=True
)


# ============================================================
# 9. INSPECT ALL FOLD-LEVEL RESULTS
# ============================================================

print("\nAll Gradient Boosting tuning results:")

display(
    gb_tuning_results
)


# ============================================================
# 10. SUMMARIZE EACH PARAMETER CONFIGURATION
# ============================================================

gb_tuning_summary = (
    gb_tuning_results
    .groupby(
        [
            "n_estimators",
            "learning_rate",
            "max_depth",
        ]
    )
    .agg(
        mean_MAE=("MAE", "mean"),
        std_MAE=("MAE", "std"),
        mean_RMSE=("RMSE", "mean"),
        std_RMSE=("RMSE", "std"),
        mean_R2=("R2", "mean"),
        std_R2=("R2", "std"),
    )
    .reset_index()
    .sort_values(
        "mean_MAE"
    )
    .reset_index(drop=True)
)

print(
    "\nGradient Boosting tuning summary:"
)

display(
    gb_tuning_summary
)


# ============================================================
# 11. IDENTIFY BEST CONFIGURATION
# ============================================================

best_params = (
    gb_tuning_summary
    .iloc[0]
)

best_n_estimators = int(
    best_params["n_estimators"]
)

best_learning_rate = float(
    best_params["learning_rate"]
)

best_max_depth = int(
    best_params["max_depth"]
)

print("\nBest Gradient Boosting configuration:")

print(
    "n_estimators:",
    best_n_estimators
)

print(
    "learning_rate:",
    best_learning_rate
)

print(
    "max_depth:",
    best_max_depth
)

print(
    "Mean MAE:",
    best_params["mean_MAE"]
)

print(
    "MAE standard deviation:",
    best_params["std_MAE"]
)

print(
    "Mean RMSE:",
    best_params["mean_RMSE"]
)

print(
    "Mean R2:",
    best_params["mean_R2"]
)


# ============================================================
# 12. VIEW BEST MODEL PERFORMANCE BY YEAR
# ============================================================

best_model_name = (
    f"GB_n{best_n_estimators}"
    f"_lr{best_learning_rate}"
    f"_d{best_max_depth}"
)

best_fold_results = (
    gb_tuning_results[
        gb_tuning_results["model"]
        == best_model_name
    ]
    .copy()
)

print(
    "\nBest configuration performance "
    "by validation year:"
)

display(
    best_fold_results[
        [
            "validation_year",
            "train_size",
            "validation_size",
            "MAE",
            "RMSE",
            "R2",
        ]
    ]
)


# ============================================================
# 13. SAVE RESULTS
# ============================================================

gb_tuning_results.to_csv(
    RESULTS_DIR
    / "gradient_boosting_tuning_folds.csv",
    index=False
)

gb_tuning_summary.to_csv(
    RESULTS_DIR
    / "gradient_boosting_tuning_summary.csv",
    index=False
)

best_fold_results.to_csv(
    RESULTS_DIR
    / "best_gradient_boosting_cv_results.csv",
    index=False
)

print("\nResults saved to:")
print(RESULTS_DIR)


from sklearn.ensemble import GradientBoostingRegressor
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# Lock the hyperparameters selected using temporal CV
best_params = {
    "n_estimators": 100,
    "learning_rate": 0.05,
    "max_depth": 2,
    "random_state": 42
}

final_predictions = []


# ============================================================
# TEST YEAR 2022
# Train only through 2020
# 2021 acts as the one-year gap
# ============================================================

train_2022 = modeling_table[
    modeling_table["release_year"] <= 2020
].copy()

test_2022 = modeling_table[
    modeling_table["release_year"] == 2022
].copy()

X_train_2022 = train_2022[feature_cols]
y_train_2022 = train_2022[target]

X_test_2022 = test_2022[feature_cols]
y_test_2022 = test_2022[target]

pipeline_2022 = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        GradientBoostingRegressor(**best_params)
    )
])

pipeline_2022.fit(
    X_train_2022,
    y_train_2022
)

pred_2022 = pipeline_2022.predict(
    X_test_2022
)

results_2022 = test_2022[
    [
        "appid",
        "name",
        "release_year",
        target
    ]
].copy()

results_2022["prediction"] = pred_2022
results_2022["test_year"] = 2022

final_predictions.append(results_2022)


# ============================================================
# TEST YEAR 2023
# Train only through 2021
# 2022 acts as the one-year gap
# ============================================================

train_2023 = modeling_table[
    modeling_table["release_year"] <= 2021
].copy()

test_2023 = modeling_table[
    modeling_table["release_year"] == 2023
].copy()

X_train_2023 = train_2023[feature_cols]
y_train_2023 = train_2023[target]

X_test_2023 = test_2023[feature_cols]
y_test_2023 = test_2023[target]

pipeline_2023 = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        GradientBoostingRegressor(**best_params)
    )
])

pipeline_2023.fit(
    X_train_2023,
    y_train_2023
)

pred_2023 = pipeline_2023.predict(
    X_test_2023
)

results_2023 = test_2023[
    [
        "appid",
        "name",
        "release_year",
        target
    ]
].copy()

results_2023["prediction"] = pred_2023
results_2023["test_year"] = 2023

final_predictions.append(results_2023)


# ============================================================
# COMBINE FINAL HOLDOUT PREDICTIONS
# ============================================================

final_predictions = pd.concat(
    final_predictions,
    ignore_index=True
)

y_true = final_predictions[target]
y_pred = final_predictions["prediction"]


# ============================================================
# FINAL METRICS
# ============================================================

final_mae = mean_absolute_error(
    y_true,
    y_pred
)

final_rmse = np.sqrt(
    mean_squared_error(
        y_true,
        y_pred
    )
)

final_r2 = r2_score(
    y_true,
    y_pred
)

print("FINAL HOLDOUT RESULTS")
print("---------------------")
print("Number of games:", len(final_predictions))
print("MAE:", final_mae)
print("RMSE:", final_rmse)
print("R2:", final_r2)

final_year_results = []

for year in [2022, 2023]:

    year_df = final_predictions[
        final_predictions["test_year"] == year
    ]

    year_mae = mean_absolute_error(
        year_df[target],
        year_df["prediction"]
    )

    year_rmse = np.sqrt(
        mean_squared_error(
            year_df[target],
            year_df["prediction"]
        )
    )

    year_r2 = r2_score(
        year_df[target],
        year_df["prediction"]
    )

    final_year_results.append({
        "year": year,
        "n_games": len(year_df),
        "MAE": year_mae,
        "RMSE": year_rmse,
        "R2": year_r2
    })

final_year_results = pd.DataFrame(
    final_year_results
)

final_year_results

final_predictions.to_csv(
    RESULTS_DIR / "final_holdout_predictions.csv",
    index=False
)

final_year_results.to_csv(
    RESULTS_DIR / "final_holdout_results_by_year.csv",
    index=False
)

from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import RandomForestRegressor


def evaluate_final_holdout_model(
    model_name,
    model,
    modeling_table,
    feature_cols,
    target
):
    predictions_all = []

    # Same availability logic as our final Gradient Boosting test
    test_settings = [
        {
            "test_year": 2022,
            "train_end": 2020
        },
        {
            "test_year": 2023,
            "train_end": 2021
        }
    ]

    for setting in test_settings:

        train_df = modeling_table[
            modeling_table["release_year"]
            <= setting["train_end"]
        ].copy()

        test_df_year = modeling_table[
            modeling_table["release_year"]
            == setting["test_year"]
        ].copy()

        X_train = train_df[feature_cols]
        y_train = train_df[target]

        X_test = test_df_year[feature_cols]
        y_test = test_df_year[target]

        pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "model",
                model
            )
        ])

        pipeline.fit(
            X_train,
            y_train
        )

        predictions = pipeline.predict(
            X_test
        )

        result = pd.DataFrame({
            "model": model_name,
            "test_year": setting["test_year"],
            "actual": y_test.values,
            "prediction": predictions
        })

        predictions_all.append(result)

    predictions_all = pd.concat(
        predictions_all,
        ignore_index=True
    )

    mae = mean_absolute_error(
        predictions_all["actual"],
        predictions_all["prediction"]
    )

    rmse = np.sqrt(
        mean_squared_error(
            predictions_all["actual"],
            predictions_all["prediction"]
        )
    )

    r2 = r2_score(
        predictions_all["actual"],
        predictions_all["prediction"]
    )

    return {
        "model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    }
    

    from sklearn.dummy import DummyRegressor
from sklearn.linear_model import LinearRegression
from sklearn.ensemble import (
    RandomForestRegressor,
    GradientBoostingRegressor
)
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.metrics import (
    mean_absolute_error,
    mean_squared_error,
    r2_score
)

# ============================================================
# FINAL HOLDOUT MODEL COMPARISON
# ============================================================

def evaluate_final_holdout_model(
    model_name,
    model,
    modeling_table,
    feature_cols,
    target
):
    predictions_all = []

    # Preserve the same one-year gap used in temporal CV
    test_settings = [
        {
            "test_year": 2022,
            "train_end": 2020
        },
        {
            "test_year": 2023,
            "train_end": 2021
        }
    ]

    for setting in test_settings:

        train_df = modeling_table[
            modeling_table["release_year"]
            <= setting["train_end"]
        ].copy()

        test_df_year = modeling_table[
            modeling_table["release_year"]
            == setting["test_year"]
        ].copy()

        X_train = train_df[feature_cols]
        y_train = train_df[target]

        X_test = test_df_year[feature_cols]
        y_test = test_df_year[target]

        pipeline = Pipeline([
            (
                "imputer",
                SimpleImputer(strategy="median")
            ),
            (
                "model",
                model
            )
        ])

        pipeline.fit(
            X_train,
            y_train
        )

        predictions = pipeline.predict(
            X_test
        )

        result = pd.DataFrame({
            "model": model_name,
            "test_year": setting["test_year"],
            "actual": y_test.to_numpy(),
            "prediction": predictions
        })

        predictions_all.append(result)

    predictions_all = pd.concat(
        predictions_all,
        ignore_index=True
    )

    mae = mean_absolute_error(
        predictions_all["actual"],
        predictions_all["prediction"]
    )

    rmse = np.sqrt(
        mean_squared_error(
            predictions_all["actual"],
            predictions_all["prediction"]
        )
    )

    r2 = r2_score(
        predictions_all["actual"],
        predictions_all["prediction"]
    )

    return {
        "model": model_name,
        "MAE": mae,
        "RMSE": rmse,
        "R2": r2
    }


# ============================================================
# RUN ALL FINAL HOLDOUT MODELS
# ============================================================

final_model_comparison = []

final_model_comparison.append(
    evaluate_final_holdout_model(
        model_name="Dummy Median",
        model=DummyRegressor(
            strategy="median"
        ),
        modeling_table=modeling_table,
        feature_cols=feature_cols,
        target=target
    )
)

final_model_comparison.append(
    evaluate_final_holdout_model(
        model_name="Linear Regression",
        model=LinearRegression(),
        modeling_table=modeling_table,
        feature_cols=feature_cols,
        target=target
    )
)

final_model_comparison.append(
    evaluate_final_holdout_model(
        model_name="Random Forest",
        model=RandomForestRegressor(
            n_estimators=200,
            random_state=42
        ),
        modeling_table=modeling_table,
        feature_cols=feature_cols,
        target=target
    )
)

final_model_comparison.append(
    evaluate_final_holdout_model(
        model_name="Tuned Gradient Boosting",
        model=GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.05,
            max_depth=2,
            random_state=42
        ),
        modeling_table=modeling_table,
        feature_cols=feature_cols,
        target=target
    )
)


# ============================================================
# CREATE COMPARISON TABLE
# ============================================================

final_model_comparison = (
    pd.DataFrame(final_model_comparison)
    .sort_values("MAE")
    .reset_index(drop=True)
)

final_model_comparison


final_model_comparison.to_csv(
    RESULTS_DIR / "final_model_comparison.csv",
    index=False
)

print("Saved final model comparison.")

final_model_comparison

# ============================================================
# FEATURE IMPORTANCE
# ============================================================

import matplotlib.pyplot as plt

# Fit the selected Gradient Boosting model on all development
# data through 2021 for interpretation.
X_development = development_df[feature_cols]
y_development = development_df[target]

final_interpretability_pipeline = Pipeline([
    (
        "imputer",
        SimpleImputer(strategy="median")
    ),
    (
        "model",
        GradientBoostingRegressor(
            n_estimators=100,
            learning_rate=0.05,
            max_depth=2,
            random_state=42
        )
    )
])

final_interpretability_pipeline.fit(
    X_development,
    y_development
)

# Extract fitted Gradient Boosting model
fitted_gb = final_interpretability_pipeline.named_steps["model"]

feature_importance = pd.DataFrame({
    "feature": feature_cols,
    "importance": fitted_gb.feature_importances_
})

feature_importance = (
    feature_importance
    .sort_values(
        "importance",
        ascending=False
    )
    .reset_index(drop=True)
)

feature_importance

INTERPRETABILITY_DIR = (
    RESULTS_DIR / "interpretability"
)

INTERPRETABILITY_DIR.mkdir(
    parents=True,
    exist_ok=True
)

feature_importance.to_csv(
    INTERPRETABILITY_DIR
    / "gradient_boosting_feature_importance.csv",
    index=False
)

top_features = feature_importance.head(15)

plt.figure(figsize=(10, 7))

plt.barh(
    top_features["feature"][::-1],
    top_features["importance"][::-1]
)

plt.xlabel("Feature importance")
plt.ylabel("Feature")
plt.title(
    "Gradient Boosting Feature Importance"
)

plt.tight_layout()

plt.savefig(
    INTERPRETABILITY_DIR
    / "gradient_boosting_feature_importance.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
